```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assisant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages):
    message = client.messages.create(
    model = model,
    max_tokens=200,
    messages=messages,
)
    
    return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

#Add in the initial user question of "Define who is mil mascaras in one sentence"
messages.append({"role": "user", "content": "Define who is mil mascaras in one sentence"})

messages

# Call the API directly to get an answer
message = client.messages.create(
	model=model,
	max_tokens=200,
	messages=messages,
)

answer = message.content[0].text

answer

# Take the answer and add it as an assistant message into our list 
messages.append({"role": "assistant", "content": answer})

#Add in the users follow up question
messages.append({"role": "user", "content": "Write another sentence"})

# Call the API again with the updated list of messages
message = client.messages.create(
	model=model,
	max_tokens=200,
	messages=messages,
)

answer = message.content[0].text
answer
```
